## Conclusion

In this assignment, an Artificial Neural Network (ANN) was developed for Credit Card Fraud Detection. The dataset was cleaned and preprocessed, missing values and duplicate records were checked, and the class imbalance was handled using SMOTE. Feature scaling was also applied before training the models.

Three ANN models were experimented with using different activation functions and optimizers. **Model 1, using ReLU activation with the Adam optimizer, achieved the best overall performance**, with an accuracy of **99.93%** and an F1-score of **74.0%**. Model 3 achieved the highest recall of **80.3%**, showing better fraud detection capability, but Model 1 provided the best overall balance between precision and recall.

Therefore, **Model 1 (ReLU + Adam)** was selected as the best-performing model for this fraud detection task. The results also demonstrate that accuracy alone is not sufficient for evaluating an imbalanced fraud detection dataset; **precision, recall, and F1-score are important metrics for assessing the model's effectiveness.**
